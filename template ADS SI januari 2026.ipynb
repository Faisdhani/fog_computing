```python
# 1. Drop kolom yang tidak relevan - buka pagar jika anda ingin pakai
# print("\n1. DROP KOLOM YANG TIDAK RELEVAN:")
# columns_to_drop = ['isi saya']
# df_clean = df_clean.drop(columns=columns_to_drop)
# print(f"   Kolom yang di-drop: {columns_to_drop}")
# print(f"   Jumlah kolom setelah drop: {df_clean.shape[1]}")
```

```python
# 2. Outlier detection
print("\n2. OUTLIER DETECTION (IQR Method):")
numeric_cols_clean = df_clean.select_dtypes(include=[np.number]).columns.tolist()
outlier_summary = {}
for col in numeric_cols_clean:
    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df_clean[(df_clean[col] < lower_bound) | (df_clean[col] > upper_bound)]
    outlier_count = len(outliers)
    outlier_percentage = (outlier_count / len(df_clean)) * 100

    outlier_summary[col] = {
        'count': outlier_count,
        'percentage': outlier_percentage,
        'lower_bound': lower_bound,
        'upper_bound': upper_bound
    }

    if outlier_count > 0:
        print(f"\n   {col}:")
        print(f"   - Bounds: [{lower_bound:.2f}, {upper_bound:.2f}]")
        print(f"   - Outliers: {outlier_count} ({outlier_percentage:.2f}%)")

# Visualisasi outliers
fig, axes = plt.subplots(2, 3, figsize=(15, 10))
axes = axes.ravel()

for idx, col in enumerate(numeric_cols_clean):
    if idx < len(axes):
        axes[idx].boxplot(df_clean[col].dropna(), vert=True)
        axes[idx].set_title(f'{col}')
        axes[idx].set_ylabel('Value')
        axes[idx].grid(True, alpha=0.3)

plt.suptitle('Boxplot untuk Deteksi Outlier', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig(FIGURES_PATH + '06_outlier_detection.png', dpi=300, bbox_inches='tight')
print("\n Visualisasi disimpan: 06_outlier_detection.png")
plt.show()
```

```python
# 3. Handle missing values
print("\n3. HANDLE MISSING VALUES:")

print(f"\n   Total missing values sebelum imputasi: {df_clean.isnull().sum().sum()}\n")

# Imputasi numerik dengan median
for col in numeric_cols_clean:
    if df_clean[col].isnull().sum() > 0:
        median_val = df_clean[col].median()
        df_clean[col].fillna(median_val, inplace=True)
        print(f"   {col}: Diimputasi dengan median = {median_val}")

# Imputasi kategorikal dengan modus
categorical_cols_clean = df_clean.select_dtypes(include=['object']).columns.tolist()
for col in categorical_cols_clean:
    if df_clean[col].isnull().sum() > 0:
        mode_val = df_clean[col].mode()[0]
        before_na = df_clean[col].isnull().sum()
        df_clean[col].fillna(mode_val, inplace=True)
        print(f"   {col}: {before_na} missing values diimputasi dengan modus = '{mode_val}'")

print(f"\n   Total missing values setelah imputasi: {df_clean.isnull().sum().sum()}")
```

```python
# 4. Hapus fitur bernilai konstan
print("\n4. DROP CONSTANT FEATURES:\n")
print(f"   Jumlah kolom sebelum cek kolom fitur yang bernilai konstan: {df_clean.shape[1]}")
df_clean = df_clean.loc[:,df_clean.apply(pd.Series.nunique) != 1]
print(f"   Jumlah kolom setelah cek kolom fitur yang bernilai konstan: {df_clean.shape[1]}")
```

```python
# 5. Drop Duplicates
# Hapus duplikat jika ada
print("\n5. DROP DUPLICATES:")
duplicates = df_clean.duplicated().sum()
if duplicates > 0:
    df_clean = df_clean.drop_duplicates(keep='first')
    print(f"\n Removed {duplicates} duplicate records")
else:
    print(f"\n Tidak ada duplicate records")

print(f"\nJumlah records setelah cleaning: {len(df_clean)}")
```
